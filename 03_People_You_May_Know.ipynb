```python
import json 
def load(file):
    with open(file,"r") as f:
        return json.load(f)
    
def find_people_u_may_know(user_id,data):
   user_frnds= {}
   for user in data["users"]:
        user_frnds[user["id"]]=set(user["friends"])
   
   if user_id not in user_frnds:
     return []
    
   direct_frnd = user_frnds[user_id] 
   suggestion = {}
    
   for frnd in direct_frnd:
    for mutual in user_frnds[frnd]:
        if mutual!=user_id and mutual not in direct_frnd:
            suggestion[mutual]=suggestion.get(mutual,0)+1
            
    sorted_suggestion = sorted(suggestion.items(), key = lambda x: x[1], reverse = True)
    return [user_id for user_id,_ in sorted_suggestion]

data = load("massive_data.json")
user_id= 16
recc = find_people_u_may_know(user_id,data)
print(f"for user id {user_id} recodamended user id is{recc}")
```
